Which carriers have the highest average transit time (TPT), and how do they compare to the agreed service level?


```python
FreightRates.head()
```

```python
OrderList.head()
```

```python
temp = FreightRates.join(OrderList, lsuffix='_left', rsuffix='_right')
temp.columns
```

```python
temp.groupby(["Carrier_left"])["TPT"].mean().sort_values(ascending = False)
```

```python
actual = (
    OrderList
    .groupby("Carrier")
    .agg(
        total_shipments=("Order ID", "count"),
        avg_actual_tpt=("TPT", "mean")
    )
    .round(2)
    .reset_index()
)
actual
```

What is the total freight cost per carrier, and which carrier is the most expensive per unit of weight shipped?

```python
temp2 = temp.groupby(["Carrier_left"])["Weight"].sum().sort_values(ascending = False)
```

```python
temp3 = temp.groupby(["Carrier_left"])["rate"].sum().sort_values(ascending = False)
temp3
```

```python
temp2*temp3
```

Which carriers have the highest rate of late shipments, and by how many days on average?

```python
actual = (
    OrderList
    .groupby("Carrier")
    .agg(
        total_shipments=("Order ID", "count"),
        avg_actual_tpt=("TPT", "mean")
    )
    .round(2)
    .reset_index()
)
actual
```

```python
OrderList.columns
```

```python
OrderList.groupby(["Carrier"]).agg(
    totalLateShipments = ("Ship Late Day count","sum"),
    totalShipments = ("Ship Late Day count","count"),
    avgLateShipments = ("Ship Late Day count","mean")
)
```

Which plants are consistently shipping orders early, and what is the cost implication of excess early shipments?

```python
OrderList.groupby(["Plant Code"]).agg(
    totalEarlyShipments = ("Ship ahead day count","mean")
).sort_values(ascending = False, by = ["totalEarlyShipments"]).reset_index()
```
